
Inicialización: **Paso a paso del algoritmo Hill Climbing** <br>

Se selecciona un estado inicial arbitrario (puede ser aleatorio o definido por el usuario). <br>
**Generación de vecinos:** <br>

Se generan pequeños cambios en el estado actual para obtener un conjunto de estados vecinos. <br>

**Evaluación:** <br>

Se calcula el valor de una función objetivo para cada estado vecino.<br>

*   Selección del mejor vecino: Se elige el vecino con el mejor valor de la función objetivo.
*   Si el mejor vecino es mejor que el estado actual, se avanza a ese nuevo estado

**Condición de Parada** <br>
*   Si no hay un vecino mejor (se alcanza un óptimo local) o se cumple algún otro criterio de finalización (como un número máximo de iteraciones), el algoritmo se detiene.

```python
import random

def funcion_objetivo(x):
    """Función que queremos maximizar"""
    return - (x - 3) ** 2 + 9

def hill_climbing():
    """Implementación básica del algoritmo Hill Climbing"""

    # Paso 1: Elegimos un estado inicial aleatorio en el rango [-10, 10]
    estado_actual = random.uniform(-10, 10)
    max_iteraciones = 100  # Límite de iteraciones para evitar bucles infinitos
    paso = 0.5  # Tamaño del cambio (vecinos)

    for _ in range(max_iteraciones):
        # Paso 2: Generamos vecinos (izquierda y derecha)
        vecino_derecha = estado_actual + paso
        vecino_izquierda = estado_actual - paso

        # Evaluamos los valores de la función objetivo
        valor_actual = funcion_objetivo(estado_actual)
        valor_derecha = funcion_objetivo(vecino_derecha)
        valor_izquierda = funcion_objetivo(vecino_izquierda)

        # Paso 3: Escogemos el mejor vecino
        if valor_derecha > valor_actual:
            estado_actual = vecino_derecha
        elif valor_izquierda > valor_actual:
            estado_actual = vecino_izquierda
        else:
            # Paso 4: Detenemos si no hay mejora
            break

    return estado_actual, funcion_objetivo(estado_actual)

# Ejecutamos el algoritmo
mejor_x, mejor_y = hill_climbing()
print(f"Mejor solución encontrada: x = {mejor_x:.2f}, f(x) = {mejor_y:.2f}")
```
