```python
# Grafo representado como diccionario de adyacencia con pesos
from copy import copy

def llenar_jarra(jarras: list[int, int], idx: int):
  jarras[idx] = 3 if idx else 4
  return jarras

def vaciar_jarra_idx(jarras: list[int, int], idx: int):
  jarras[idx] = 0
  return jarras
 
def vaciar_en_idx(jarras: list[int,int], idx: int):
    capacidad_destino = 3 if idx else 4
    cantidad = min(jarras[1-idx], capacidad_destino - jarras[idx])
    jarras[1 - idx] -= cantidad
    jarras[idx] += cantidad
    return jarras

def anchura(inicio):
  vistos = {tuple(inicio)}
  grafo = [inicio]

  while grafo:
    padre = grafo.pop(0)
    print("> padre", padre)


    hijos_totales = []
    for i in range(2):
      llenar = llenar_jarra(copy(padre), i)
      dejar = vaciar_en_idx(copy(padre), i)
      mover = vaciar_en_idx(copy(padre), 1-i)
      vaciar = vaciar_jarra_idx(copy(padre), i)

      hijos = [llenar, dejar, mover, vaciar]
      hijos = [[x, y] for x, y in hijos if (x, y) not in vistos]
      hijos_totales += hijos
      vistos = vistos | {tuple(hijo) for hijo in hijos}

      hijo_solucion = [[x, y] for x, y in hijos if x == 2]
      if hijo_solucion:
        print("< solucion!", *hijo_solucion)
        return
      
      grafo += hijos
    print(f"  - {hijos_totales=}")
    print(f"  > {grafo=}\n")


inicio = [0, 0]
anchura(inicio)
```
