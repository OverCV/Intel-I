**Implementacion Primera Red Neuronal**

image.png

```python
import tensorflow as tf
import numpy as np
```

```python
celsius = np.array([-40,-10,0,8,15,22,38], dtype=float)
farenheit = np.array([-40,14,32,46,59,72,100], dtype=float)
```

image.png

```python
capa = tf.keras.layers.Dense(units=1)
modelo = tf.keras.Sequential()
modelo.add(tf.keras.Input(shape=(1,)))
modelo.add(capa)
```

```python
modelo.compile(optimizer=tf.keras.optimizers.Adam(0.1),
               loss="mean_squared_error"
)
```

```python
print("Etapa de Entrenamiento")
historial = modelo.fit(celsius, farenheit,epochs=1000,verbose=False)
print("Red Neuronal Entrenada")
```

```python
import matplotlib.pyplot as plt
plt.xlabel("# Epoca")
plt.ylabel("Magnitud de la perdida")
plt.plot(historial.history["loss"])
```

```python
# HAcer la predicción a partir del modelo generado
print("Listos para predecir")
resultado = modelo.predict([100])
print(resultado)
```

```python
## ver los resultados del entrenamiento
print("Valores internos del aprendizaje")
print(capa.get_weights())
```
