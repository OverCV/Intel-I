# Algoritmo Minimax para Tres en Raya

```python
from typing import List, Optional, Tuple

class SYM:
    X = 'X'
    O = 'O'
    SPACE = ' '
Board = List[str]

def print_board(board: Board):
    print("\n".join([" | ".join(board[i:i+3]) for i in range(0, 9, 3)]))
    print()

def winner(board: Board) -> Optional[str]:
    lines = [
        [0, 1, 2], [3, 4, 5], [6, 7, 8],
        [0, 3, 6], [1, 4, 7], [2, 5, 8],
        [0, 4, 8], [2, 4, 6]
    ]
    for a, b, c in lines:
        if board[a] == board[b] == board[c] and board[a] != SYM.SPACE:
            return board[a]
    return None

def is_full(board: Board) -> bool:
    return ' ' not in board

def minimax(board: Board, is_max: bool) -> Tuple[int, Optional[int]]:
    w = winner(board)
    if w == SYM.X:
        return 1, None
    if w == SYM.O:
        return -1, None
    if is_full(board):
        return 0, None

    best_move = None
    if is_max:
        best_score = -float('inf')
        for i in range(9):
            if board[i] == SYM.SPACE:
                board[i] = SYM.X
                score, _ = minimax(board, False)
                board[i] = SYM.SPACE
                if score > best_score:
                    best_score = score
                    best_move = i
    else:
        best_score = float('inf')
        for i in range(9):
            if board[i] == ' ':
                board[i] = 'O'
                score, _ = minimax(board, True)
                board[i] = ' '
                if score < best_score:
                    best_score = score
                    best_move = i
    return best_score, best_move
```

```python
board = ['X', 'O', 'X',
         'O', 'X', ' ',
         ' ', ' ', 'O']

print_board(board)
_, move = minimax(board, True)
print(f"Mejor movimiento para 'X': {move}")
```
